1. **Contador de Linhas**
	- Escreva um programa que receba o nome de um arquivo de texto e exiba o número total de linhas desse arquivo.


```python
arquivo = str(input("Digite o nome do arquivo"))


def contador_linhas(arquivo):
    with open(f'{arquivo}.txt', 'r') as arq:
        linhas = arq.readlines()
        contador = 0

        for linha in linhas:
            contador += 1

        print(f"O arquivo tem {contador} linhas")


try:
    contador_linhas(arquivo)
except FileNotFoundError:
    print("Arquivo não existe")
```

2. **Filtro de Linhas por Palavra**
	- Crie um utilitário que receba um arquivo de texto e uma palavra, e salve em um novo arquivo apenas as linhas que contêm essa palavra.

```python
arquivo = str(input("Digite o nome do arquivo"))
palavra = str(input("Digite uma palavra"))

with open(f"{arquivo}.txt", 'r') as arq:
    linhas = arq.readlines()
    pesquisa = []

    for linha in linhas:
        if palavra in linha:
           pesquisa.append(linha)

    novo_arquivo = open("novo_arquivo.txt", 'w')
    novo_arquivo.writelines(pesquisa)
```

3. **Conversor de CSV para TXT**
	- Implemente um programa que leia um arquivo CSV e salve os dados em um arquivo de texto simples, um registro por linha.

```python
import csv

lista = [
	['nome', 'idade', 'nascimento'],
    ['leo', '29', '1995'],
    ['lucas', '21', '2000'],
    ['gustavo', '40', '1980'],
    ['will', '35', '1991']
]

with open('arquivo.csv', 'w', newline="", encoding='utf-8') as file:
	escritor = csv.writer(file)
	escritor.writerows(lista)

with open('arquivo.csv', 'r', encoding='utf-8') as file:
    leitor = csv.reader(file)
    lista = []
    for linha in leitor:
        lista.append(linha)

with open("conversao.txt", 'w') as arquivo:
    for linha in lista:
        arquivo.write(f"{linha} \n")
    
```
